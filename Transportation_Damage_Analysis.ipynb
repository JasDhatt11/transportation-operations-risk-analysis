```python
result = con.execute("""
    SELECT
        Route_Risk,
        Loading_Method,
        COUNT(*) AS incidents,
        ROUND(AVG(Handling_Count), 2) AS avg_handling_count,
        ROUND(AVG(Damage_Cost), 2) AS avg_damage_cost,
        ROUND(AVG(Damage_Cost_Rate), 4) AS avg_damage_rate,
        ROUND(AVG(Delay_Hours), 2) AS avg_delay_hours
    FROM read_csv_auto('trucking_incident_dataset_1000.csv')
    GROUP BY Route_Risk, Loading_Method
    HAVING COUNT(*) >= 10
    ORDER BY avg_damage_cost DESC
""").df()

result
```

```python
result = con.execute("""
    SELECT
        Route_Risk,
        Loading_Method,
        COUNT(*) AS incidents,
        ROUND(AVG(Shipment_Weight_KG), 2) AS avg_weight_kg,
        ROUND(AVG(Shipment_Value), 2) AS avg_shipment_value,
        ROUND(AVG(Distance_KM), 2) AS avg_distance_km,
        ROUND(AVG(Damage_Cost), 2) AS avg_damage_cost
    FROM read_csv_auto('trucking_incident_dataset_1000.csv')
    GROUP BY Route_Risk, Loading_Method
    HAVING COUNT(*) >= 10
    ORDER BY avg_damage_cost DESC
""").df()

result
```

```python
result = con.execute("""
    SELECT
        Facility,
        Route_Risk,
        Loading_Method,
        COUNT(*) AS incidents,
        ROUND(AVG(Damage_Cost), 2) AS avg_damage_cost,
        ROUND(AVG(Damage_Cost_Rate), 4) AS avg_damage_rate,
        ROUND(AVG(Handling_Count), 2) AS avg_handling_count,
        ROUND(AVG(Delay_Hours), 2) AS avg_delay_hours
    FROM read_csv_auto('trucking_incident_dataset_1000.csv')
    WHERE Route_Risk = 'Medium'
      AND Loading_Method = 'Conveyor'
    GROUP BY Facility, Route_Risk, Loading_Method
    HAVING COUNT(*) >= 5
    ORDER BY avg_damage_cost DESC
""").df()

result
```

```python
result = con.execute("""
    SELECT
        Facility,
        Damage_Type,
        Damage_Severity,
        COUNT(*) AS incidents,
        ROUND(AVG(Damage_Cost), 2) AS avg_damage_cost,
        ROUND(SUM(Damage_Cost), 2) AS total_damage_cost
    FROM read_csv_auto('trucking_incident_dataset_1000.csv')
    WHERE Route_Risk = 'Medium'
      AND Loading_Method = 'Conveyor'
      AND Facility IN ('Brampton Cross-Dock',
                       'Mississauga Cross-Dock')
    GROUP BY Facility, Damage_Type, Damage_Severity
    ORDER BY Facility, total_damage_cost DESC
""").df()

result
```

```python
result = con.execute("""
    SELECT
        Facility,
        Route,
        Route_Risk,
        Weather,
        Loading_Method,
        Vehicle_Type,
        Driver_Experience,
        Handling_Count,
        Shipment_Weight_KG,
        Shipment_Value,
        Damage_Type,
        Damage_Severity,
        Damage_Cost,
        Delay_Hours
    FROM read_csv_auto('trucking_incident_dataset_1000.csv')
    WHERE Damage_Severity = 'Critical'
    ORDER BY Damage_Cost DESC
""").df()

result
```

```python
result = con.execute("""
    SELECT
        Facility,
        COUNT(*) AS incidents,
        SUM(Damage_Cost) AS total_damage_cost,
        ROUND(AVG(Damage_Cost), 2) AS avg_damage_cost,
        SUM(Claim_Amount) AS total_claim_amount,
        ROUND(AVG(Delay_Hours), 2) AS avg_delay_hours
    FROM read_csv_auto('trucking_incident_dataset_1000.csv')
    GROUP BY Facility
    ORDER BY total_damage_cost DESC
""").df()

result
```

```python
result = con.execute("""
    SELECT
        Facility,
        COUNT(*) AS incidents,
        ROUND(SUM(Damage_Cost), 2) AS total_damage_cost,
        ROUND(SUM(Claim_Amount), 2) AS total_claim_amount,
        ROUND(
            SUM(Claim_Amount) / NULLIF(SUM(Damage_Cost), 0) * 100,
            2
        ) AS claim_recovery_pct
    FROM read_csv_auto('trucking_incident_dataset_1000.csv')
    GROUP BY Facility
    ORDER BY claim_recovery_pct DESC
""").df()

result
```
